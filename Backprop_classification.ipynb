Back Propagation - Using Keras

```python
import numpy as np
import pandas as pd
df=pd.DataFrame([[8,8,1],[6,7,0],[8,10,1],[9,12,1]], columns=['cgpa','profile_score','lpa'])
df
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense

model=Sequential()
model.add(Dense(2,activation='sigmoid',input_dim=2))
model.add(Dense(1,activation='sigmoid'))
model.summary()
```

```python
new_weights=[np.array([[0.1, 0.1 ],
[ 0.1,  0.1]], dtype=np.float32),
np.array([0., 0.], dtype=np.float32),
np.array([[0.1  ],
[0.1]], dtype=np.float32),
np.array([0.], dtype=np.float32)]
model.set_weights(new_weights)
model.get_weights()
```

```python
from keras.optimizers import Adam
optimizer=Adam(learning_rate=0.001)
model.compile(loss='binary_crossentropy',optimizer=optimizer)
```

```python
model.fit(df.iloc[:,0:-1].values,df.iloc[:,-1].values,epochs=100,verbose=1,batch_size=1)
```

Back Propagation - Without using Keras

```python
import numpy as np
import pandas as pd
df=pd.DataFrame([[8,8,1],[6,7,0],[8,10,1],[9,12,1]], columns=['cgpa','profile_score','lpa'])
df
```

```python
def initialize_parameters(layer_dims):
  #np.random.seed(3)
  parameters={}
  L=len(layer_dims)
  for i in range(1,L):
    parameters['w'+str(i)]=np.ones((layer_dims[i-1],layer_dims[i]))*0.1
    parameters['b'+str(i)]=np.zeros((layer_dims[i],1))
  return parameters
```

```python
def sigmoid(z):
  return 1/(1+np.exp(-z))
```

```python
def linear_forward(A_prev,w,b):
  A=np.dot(w.T,A_prev)+b
  return sigmoid(A)
```

```python
def L_layer_forward(X, parameters):
  A=X
  L=len(parameters)//2
  for i in range(1,L+1):
    A_prev=A
    w1=parameters['w'+str(i)]
    b1=parameters['b'+str(i)]
    A=linear_forward(A_prev,w1,b1)
  return A,A_prev
```

```python
X=df[['cgpa','profile_score']].values[0].reshape(2,1)
y=df[['lpa']].values[0][0]
parameters=initialize_parameters([2,2,1])
y_hat,A1=L_layer_forward(X, parameters)
y_hat=y_hat[0][0]
X,y,parameters,y_hat,A1
```

```python
def update_parameters(parameters,y,y_hat,A1,X,learn_rate):
  parameters['w2'][0][0]=parameters['w2'][0][0]+(learn_rate * (y-y_hat)*A1[0][0])
  parameters['w2'][1][0]=parameters['w2'][1][0]+(learn_rate * (y-y_hat)*A1[1][0])
  parameters['b2'][0][0]=parameters['b2'][0][0]+(learn_rate * (y-y_hat))
  parameters['w1'][0][0]=parameters['w1'][0][0]+(learn_rate * (y-y_hat)*parameters['w2'][0][0]*X[0][0])
  parameters['w1'][0][1]=parameters['w1'][0][1]+(learn_rate * (y-y_hat)*parameters['w2'][0][0]*X[1][0])
  parameters['b1'][0][0]=parameters['b1'][0][0]+(learn_rate * (y-y_hat)*parameters['w2'][0][0])
  parameters['w1'][1][0]=parameters['w1'][1][0]+(learn_rate * (y-y_hat)*parameters['w2'][1][0]*X[0][0])
  parameters['w1'][1][1]=parameters['w1'][1][1]+(learn_rate * (y-y_hat)*parameters['w2'][1][0]*X[1][0])
  parameters['b1'][1][0]=parameters['b1'][1][0]+(learn_rate * (y-y_hat)*parameters['w2'][1][0])
  return parameters
```

```python
# epoch implementation
parameters=initialize_parameters([2,2,1]); epochs=25001; lr=0.0001
for i in range(epochs):
  Loss=[]
  for j in range(df.shape[0]):
    X=df[['cgpa','profile_score']].values[j].reshape(2,1)
    y=df[['lpa']].values[j][0]
    y_hat,A1=L_layer_forward(X, parameters)
    y_hat=y_hat[0][0]
    update_parameters(parameters,y,y_hat,A1,X,lr)
    Loss.append((-y*np.log(y_hat))-((1-y)*np.log(1-y_hat)))
  if i%1000==0:
    print('Epoch - ',i+1,'Loss - ',np.array(Loss).mean())

parameters
```
